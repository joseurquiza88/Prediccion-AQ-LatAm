```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(10,6))
ax.axis('off')

# Cajas de texto (x,y,width,height)
boxes = [
    ("Modelos Empíricos Simples\n(Regresión Lineal Simple)\n- AOD vs PM2.5\n- Fácil implementación\n- Limitado en zonas complejas", (0.05, 0.7, 0.4, 0.2)),
    ("Modelos Multivariados y Empírico-Físicos\n(Regresión Lineal Múltiple,\nVariables: HR, capa límite)\n- Mejor estimación", (0.55, 0.7, 0.4, 0.2)),
    ("Modelos Semiparamétricos\n(GAM, efectos mixtos, GWR)\n- Captura variabilidad espacio-temporal\n- Requiere estructura previa", (0.05, 0.4, 0.4, 0.2)),
    ("Modelos de Machine Learning\n(Random Forest, XGBoost, SVR, Redes Neuronales)\n- Captura relaciones no lineales\n- Caja negra", (0.55, 0.4, 0.4, 0.2))
]

for text, (x,y,w,h) in boxes:
    rect = patches.FancyBboxPatch((x,y), w, h, boxstyle="round,pad=0.02", edgecolor="black", facecolor="#d9edf7")
    ax.add_patch(rect)
    ax.text(x + w/2, y + h/2, text, ha="center", va="center", fontsize=10)

# Flechas de evolución
ax.annotate('', xy=(0.45, 0.8), xytext=(0.55, 0.8), arrowprops=dict(arrowstyle="->", lw=2))
ax.annotate('', xy=(0.45, 0.5), xytext=(0.55, 0.5), arrowprops=dict(arrowstyle="->", lw=2))
ax.annotate('', xy=(0.25, 0.7), xytext=(0.25, 0.6), arrowprops=dict(arrowstyle="->", lw=2))
ax.annotate('', xy=(0.75, 0.7), xytext=(0.75, 0.6), arrowprops=dict(arrowstyle="->", lw=2))

plt.title("Evolución y clasificación de modelos para estimar PM2.5 a partir de AOD satelital", fontsize=14)
plt.show()
```

```python
import matplotlib.pyplot as plt

modelos = ["Empíricos", "Multivariados", "Semiparamétricos", "Machine Learning"]
complejidad = [1, 2, 3, 4]  # de baja a alta
interpretabilidad = [4, 3, 2, 1]  # de alta a baja

fig, ax = plt.subplots(figsize=(8,5))

ax.barh(modelos, complejidad, color='skyblue', label='Complejidad')
ax.barh(modelos, interpretabilidad, left=complejidad, color='lightcoral', label='Interpretabilidad')

ax.set_xlabel("Nivel")
ax.set_title("Comparación entre modelos: Complejidad y Interpretabilidad")
ax.legend()

plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

# Nodo central
G.add_node("Estimación PM2.5\ncon AOD")

# Primer nivel
modelos = ["Empíricos", "Multivariados", "Semiparamétricos", "Machine Learning"]
for m in modelos:
    G.add_edge("Estimación PM2.5\ncon AOD", m)

# Segundo nivel características (simplificado)
caracteristicas = {
    "Empíricos": ["Regresión simple", "AOD vs PM2.5"],
    "Multivariados": ["Variables físicas", "HR, capa límite"],
    "Semiparamétricos": ["GAM, GWR", "Variabilidad espacio-temporal"],
    "Machine Learning": ["Random Forest", "No linealidad"]
}
for m, attrs in caracteristicas.items():
    for attr in attrs:
        G.add_edge(m, attr)

pos = nx.spring_layout(G, k=1, seed=42)
plt.figure(figsize=(12,8))
nx.draw(G, pos, with_labels=True, node_size=3500, node_color='lightblue', font_size=10, arrowsize=20)

plt.title("Mapa conceptual de modelos para estimar PM2.5 a partir de AOD")
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

# Crear grafo dirigido
G = nx.DiGraph()

# Nodos principales con etiquetas
nodos = {
    "Modelado PM2.5 a partir de AOD": (0,4),

    "Modelos Empíricos Simples": (-2,3),
    "Modelos Multivariados y Empírico-físicos": (2,3),

    "Modelos Semiparamétricos": (-2,2),
    "Modelos de Machine Learning": (2,2),
}

# Añadir nodos al grafo
for nodo, pos in nodos.items():
    G.add_node(nodo, pos=pos)

# Añadir conexiones
edges = [
    ("Modelado PM2.5 a partir de AOD", "Modelos Empíricos Simples"),
    ("Modelado PM2.5 a partir de AOD", "Modelos Multivariados y Empírico-físicos"),
    ("Modelos Empíricos Simples", "Modelos Semiparamétricos"),
    ("Modelos Multivariados y Empírico-físicos", "Modelos de Machine Learning"),
]

G.add_edges_from(edges)

# Posiciones de nodos
pos = nx.get_node_attributes(G, 'pos')

# Dibujar
plt.figure(figsize=(12,6))
ax = plt.gca()
ax.set_title("Modelos para estimar PM2.5 a partir de AOD satelital", fontsize=16, pad=20)

# Dibujar nodos con cajas
for nodo, (x, y) in pos.items():
    bbox_props = dict(boxstyle="round,pad=0.5", fc="#cce6ff", ec="#3399ff", lw=2)
    plt.text(x, y, nodo, ha="center", va="center", size=12,
             bbox=bbox_props, wrap=True)

# Dibujar aristas con flechas
for (start, end) in edges:
    start_pos = pos[start]
    end_pos = pos[end]
    ax.annotate("",
                xy=end_pos, xycoords='data',
                xytext=start_pos, textcoords='data',
                arrowprops=dict(arrowstyle="->", color="#3366cc", lw=2),
                )

# Limpiar ejes
plt.axis('off')
plt.ylim(1, 4.5)
plt.xlim(-3, 3)

plt.show()
```
